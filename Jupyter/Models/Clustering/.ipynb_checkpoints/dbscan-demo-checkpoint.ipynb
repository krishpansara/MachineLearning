## [Visualization Tools](https://www.naftaliharris.com/blog/visualizing-dbscan-clustering/)

```python
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
```

# Demo

```python
X = np.array([[1, 2], [2, 2], [2, 3],[8, 7], [8, 8], [25, 80]])
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

## Playing with eps and min_sample

```python
db = DBSCAN(eps=3, min_samples=2)

db.fit(X)
db.labels_
```

```python
db = DBSCAN(eps=3, min_samples=3)

db.fit(X)
db.labels_
```

```python
db = DBSCAN(eps=10, min_samples=3)

db.fit(X)
db.labels_
```

# Dealing with real stuff

## Creating dataset

```python
from sklearn.datasets import make_circles

# Create a concentric circle dataset
X, _ = make_circles(n_samples=500, factor=.5, noise=.03, random_state=4)
```

## Plotting the data

```python
plt.scatter(X[:, 0], X[:, 1])
plt.title("Dataset")
plt.xlabel("Feature 0")X
plt.ylabel("Feature 1")
plt.show()
```

## Applying KNN

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2)
kmean_clust = kmeans.fit_predict(X)
```

```python
# Plotting
plt.scatter(X[:, 0], X[:, 1], c=kmean_clust, cmap='viridis', marker='o')
plt.title("KMeans Clustering of Concentric Circles")
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.show()
```

## Apply DBSCAN to the dataset

```python
dbscan = DBSCAN(eps=0.1, min_samples=5)
clusters = dbscan.fit_predict(X)
```

```python
# Plotting
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis', marker='o')
plt.title("DBSCAN Clustering of Concentric Circles")
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.show()
```
